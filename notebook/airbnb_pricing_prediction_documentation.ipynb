#### 2. Evaluation Criteria 
<hr style="width:25%;margin-left:0;">    

#### **Evaluation Metric: Mean Absolute Percentage Error (MAPE)**

The competition evaluates submissions using **Mean Absolute Percentage Error (MAPE)**, defined as:

$$MAPE = \frac{1}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right| \times 100$$

where $y_i$ is the actual nightly price, $\hat{y}_i$ is the predicted price, and $n$ is the total number of listings.

**Advantages**

- *Interpretability:* MAPE expresses prediction error as a percentage, making it immediately intuitive for non-technical stakeholders such as hosts and investors. For example, a MAPE of 20% means the model's predictions deviate from actual prices by 20% on average. This is far more meaningful than an absolute metric like MAE - a `$50` error on a `$100` listing is catastrophic, whereas the same `$50` error on a `$1,000` listing is negligible. MAPE captures this distinction naturally.

- *Scale independence:* MAPE expresses errors as percentages rather than absolute dollar values, allowing performance to be compared across listings with very different price levels. This is particularly appropriate for Airbnb price prediction, where listing prices range from budget accommodation to luxury properties. Unlike MAE, the metric accounts for the relative importance of an error compared with the listing's actual price.

- *Robustness across heterogeneous listings:* Brisbane Airbnb listings vary substantially in price, property type, and location. By evaluating prediction errors as percentages rather than dollar amounts, MAPE provides a standardised measure of forecasting performance across both budget and luxury properties.

- *Comparability:* Alignment with competition objective: Since submissions are ranked directly using MAPE, optimising this metric ensures that model development and evaluation are aligned with the competition's scoring criterion. Models with lower validation MAPE are expected to achieve better leaderboard performance.

- *Detecting overfitting and underfitting:* Although MAPE is not specifically designed to detect overfitting, comparing training and validation MAPE provides a useful diagnostic tool. A substantially lower training MAPE than validation MAPE suggests overfitting, whereas consistently high values on both datasets may indicate underfitting.

- *Alignment with competition objective:* Since competition rankings are based directly on MAPE, optimising this metric during model development ensures consistency between model evaluation and final competition performance. Models that achieve lower validation MAPE are more likely to perform well on the leaderboard.

**Limitations**

- *Division by zero:* MAPE is mathematically undefined when the actual price is zero. While this is a theoretical concern, the minimum listing
price in this dataset is $36 AUD per night, meaning division by zero cannot occur in practice. However, this remains a general limitation of MAPE when applied to other datasets.

- *Sensitivity to low-priced listings:* Listings with very low prices produce disproportionately large percentage errors even from small
absolute mistakes. For example, a `$10` error on a `$36` listing contributes 27.8% to MAPE, whereas the same `$10` error on a `$300` listing contributes only 3.3%. This asymmetry can pull the overall MAPE upward and bias models towards predicting higher prices.

- *No systematic bias detection:* Like most error metrics, MAPE does not capture whether the model systematically over- or under-predicts. A
separate bias analysis would be required to determine if predictions are consistently skewed in one direction.

- *Asymmetric penalty structure:* MAPE penalises errors on low-priced listings more heavily than identical dollar errors on expensive listings. Consequently, model optimisation may place disproportionate emphasis on accurately predicting cheaper properties while paying less attention to high-value listings.

**Despite these limitations, MAPE remains an appropriate evaluation metric because it aligns directly with the competition objective while providing an interpretable and scale-independent measure of forecasting accuracy.**

#### 3. Missing Values

```python
# Create some useful pre-define function:
def missing_summary(df, name):
    """
    Create a dataframe, then calculate the number of null and blank value.
    Only report the columns that have missing values and percentage of missing.
    
    Parameter:
    df (DataFrame)
        The dataframe used to check for missing values
    name (dataset)
        Train or test set
    
    Return:
    completed_report (DataFrame)
        The report table including the number of null, and their percentage in total
    """
    missing = df.isnull().sum()
    missing = missing[missing > 0].sort_values(ascending=False)
    pct = (missing / len(df) * 100).round(2)
    summary = pd.DataFrame({'Missing Count': missing, 'Missing (%)': pct})
    print(f"  {name}  —  {df.shape[0]} rows x {df.shape[1]} cols")
    print(f"{'='*40}")
    print(summary.to_string())
    return summary

train_missing = missing_summary(train_df, 'TRAIN')
test_missing  = missing_summary(test_df,  'TEST')


def classify_missing(pct):
    """
    Classify a column's missing-value percentage into a severity band.
    Used to group columns by how much data they are missing.

    Parameter:
    pct (float)
        The percentage of missing values in a column

    Return:
    severity (str)
        The severity label - 'Low (<5%)', 'Moderate (5-30%)', or 'High (>30%)'
    """
    if pct < 5:
        return 'Low (<5%)' 
    elif pct < 30:
        return 'Moderate (5–30%)' 
    else:
        return 'High (>30%)' 
```

```python
for name, df in [('TRAIN', train_df), ('TEST', test_df)]:
    total_cols = df.shape[1]
    cols_with_missing = df.isnull().any().sum()
    cols_no_missing = total_cols - cols_with_missing
    
    print(f"\n{'='*45}")
    print(f"  {name} — {df.shape[0]} rows x {total_cols} cols")
    print(f"{'='*45}")
    print(f"  Columns with at least 1 missing value: {cols_with_missing} ({cols_with_missing/total_cols*100:.1f}%)")
    print(f"  Columns with NO missing values:        {cols_no_missing} ({cols_no_missing/total_cols*100:.1f}%)")
```

```python
# Report missing for train and test set
if not train_missing.empty:
    train_missing['Severity'] = train_missing['Missing (%)'].apply(classify_missing)
    print("\n Missing Severity (Train) ")
    print(train_missing.groupby('Severity')['Missing Count'].count())
    print(train_missing.to_string())

if not test_missing.empty:
    test_missing['Severity'] = test_missing['Missing (%)'].apply(classify_missing)
    print("\n Missing Severity (Test) ")
    print(test_missing.groupby('Severity')['Missing Count'].count())
    print(test_missing.to_string())
```

- Missing values are classified into three types:

- The missingness appears unrelated to other observed variables and is therefore broadly consistent with MCAR. MCAR data can be removed without introducing bias. Since these columns exhibit substantial missingness and consist primarily of optional text descriptions, they will be considered for removal in Task 2.

> MAR (Missing at Random)

- All review score columns and `reviews_per_month` share exactly 437 missing values (11.70%). 
- This is because listings with `number_of_reviews = 0` have no scores to report - the missingness is explained by another observed variable, which is the definition of MAR. 
- Similarly, `host_response_rate` (7.68%) and `host_acceptance_rate` (5.38%) are likely missing for newer hosts who have not yet established a response policy. MAR data can lead to biased results if not handled properly. These will be imputed using median (numerical) or mode (categorical) in Task 2.

> MNAR (Missing Not at Random)

- No strong evidence of MNAR was found in this dataset - all missingness patterns can be explained by observable variables, as described above.

_Low missingness (<5%) - 16 columns_

Features such as `bedrooms` (0.05%), `beds` (0.08%), and `bathrooms_text` (0.16%) have very few missing values occurring essentially at random. These will be safely imputed using mean/median/mode imputation via `SimpleImputer` from scikit-learn in Task 2.

**Before selecting meaningful features, variables that provide little or no predictive value were removed.** 

- Identifier and Personal Information Variables (`ID`, `host_id`, `host_name`):

Were removed because they function primarily as identifiers rather than meaningful property characteristics. These variables do not provide information about listing quality, location, capacity, or host behaviour and therefore are unlikely to contribute useful predictive information.

- Free-text Variables (`description`, `neighborhood_overview`, `host_about`):

Were excluded because they contain unstructured textual information that would require natural language processing techniques to extract meaningful signals. Since the focus of this project is on structured data modelling, these variables were not considered further.

- Location-related Variables (`host_location`, `host_neighbourhood`):

Were removed due to inconsistent formatting, substantial missingness, and overlap with neighbourhood_cleansed, which provides a cleaner and more standardised representation of listing location.

- Administrative and Platform Metadata Variables (`source`, `host_verifications`):

Were excluded because they primarily contain platform-level information rather than direct property characteristics. In addition, host_verifications consists of complex multi-category text values that would require additional feature engineering before being suitable for modelling.

- Duplicate information column (`bathrooms_text`): 

This variable contains a textual representation of the number of bathrooms (e.g., "1 bath", "2 baths"), which duplicates the information already captured in the numerical `bathrooms` variable. Since the numeric version is more suitable for statistical analysis and modelling, `bathrooms_text` was excluded to avoid redundancy.

- Revenue Variable (`estimated_revenue_l365d`):

`estimated_revenue_l365d` is removed due to concerns regarding **target leakage**. This variable is calculated using listing price, which is the target variable that the model is trying to predict. As a result, information from the target is already embedded within the feature itself. If `estimated_revenue_l365d` were included in the model, the model could partially infer the value of `price` from a variable that was directly derived from it, rather than learning genuine relationships between price and listing characteristics such as location, property type, capacity, and host behaviour. Consequently, the model's performance may appear better than it actually is, as it is effectively being given access to information related to the answer. In contrast, `estimated_occupancy_l365d` was retained because it captures booking demand without directly incorporating the target variable.


```python
# Remove text/descriptive columns that add no signal
# NOTE: 'name' is kept here — keyword flags are extracted from it in the feature engineering step
features_to_drop = [
    'description', 'neighborhood_overview', 'host_name', 'host_id',
    'host_about', 'host_location', 'host_neighbourhood', 'host_verifications',
    'bathrooms_text'
]
train_data = raw_train_df.drop(columns=[c for c in features_to_drop if c in raw_train_df.columns])
test_data  = raw_test_df.drop( columns=[c for c in features_to_drop if c in raw_test_df.columns])
```

```python
features = train_data[num_features]
fea = pd.get_dummies(features)

fea['price'] = train_data['price_clean']

corr = fea.corr(method='pearson')

corr_no_diag = corr.copy()
np.fill_diagonal(corr_no_diag.values, 0)

price_corr = corr_no_diag['price'].abs() > 0.05
filtered_cols = price_corr[price_corr].index.tolist()

if 'price' not in filtered_cols:
    filtered_cols.append('price')

corr_filtered = corr.loc[filtered_cols, filtered_cols]

mask = np.zeros_like(corr_filtered, dtype=np.bool_)
mask[np.triu_indices_from(mask)] = True

f, ax = plt.subplots(figsize=(20, 18))

sns.heatmap(
    corr_filtered,
    annot=True,
    mask=mask,
    center=0,
    square=True,
    cmap='RdBu',
    xticklabels='auto',
)

plt.title('Correlation between the variables of train dataset')
```

#### 2. Feature Engineering 

<hr style="width:25%;margin-left:0;">   

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Feature engineering
SNAPSHOT_DATE = pd.Timestamp('2025-12-01')  # after all known scrape dates
CBD_LAT, CBD_LON = -27.4698, 153.0251 #Brisbane CBD


def haversine_cbd(lat, lon):
    R = 6371
    lat1, lon1 = np.radians(lat), np.radians(lon)
    lat2, lon2 = np.radians(CBD_LAT), np.radians(CBD_LON)
    dlat, dlon = lat2 - lat1, lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    return R * 2 * np.arcsin(np.sqrt(a))

def safe_ratio(numerator, denominator):
    """
    denominator == 0 ~ NaN (not 1). NaN will be impute later
    """
    denom = denominator.replace(0, np.nan)
    return numerator / denom

def new_features(df):
    # 1. amenities_count: number of amenities 
    am = df['amenities'].fillna('[]')
    df['amenities_count'] = np.where(am == '[]', 0, am.str.count(',') + 1)

    # 2. host_experience_days: days since host joined Airbnb
    host_since = pd.to_datetime(df['host_since'], errors='coerce')
    df['host_experience_days'] = (SNAPSHOT_DATE - host_since).dt.days.clip(lower=0)

    # 3. days_since_last_review: days since last review
    last_review = pd.to_datetime(df['last_review'], errors='coerce')
    df['days_since_last_review'] = (SNAPSHOT_DATE - last_review).dt.days

    # 4. is_commercial_host: 1 if host manages more than 1 listing
    df['is_commercial_host'] = (df['host_listings_count'] > 1).astype(int)

    # 5. beds_per_person: bed-to-guest ratio — proxy for comfort level
    df['beds_per_person'] = safe_ratio(df['beds'], df['accommodates'])

    # 6. bathroom/bedroom: bathrooms per bedroom — proxy for luxury level
    df['bath_to_bedroom_ratio'] = safe_ratio(df['bathrooms'], df['bedrooms'])

    # 7. accommodates_per_bedroom: guest capacity per bedroom — space density proxy
    df['accommodates_per_bedroom'] = safe_ratio(df['accommodates'], df['bedrooms'])

    # 8. availability_rate: proportion of year the listing is available
    df['availability_rate'] = df['availability_365'] / 365

    # 9. dist_to_cbd: Haversine distance (km) from listing to Brisbane CBD
    df['dist_to_cbd'] = haversine_cbd(df['latitude'], df['longitude'])
    
    # 10. Capture the structural 'Zero' state explicitly
    df['is_inactive_host'] = (df['estimated_occupancy_l365d'] == 0).astype(int)
    return df

# apply consistently on both set (explicit reassignment for pandas CoW safety)
train_data = new_features(train_data)
test_data  = new_features(test_data)


# -----step 2: cluster FIT ONLY TRAIN, NOT TEST
# geo_cluster: K-Means spatial cluster (fit on train only to avoid leakage)
_geo_scaler = StandardScaler().fit(train_data[['latitude', 'longitude']])
_kmeans_geo = KMeans(n_clusters=5, random_state=42, n_init=10)
_kmeans_geo.fit(_geo_scaler.transform(train_data[['latitude', 'longitude']]))

for df in [train_data, test_data]:
    labels = _kmeans_geo.predict(_geo_scaler.transform(df[['latitude', 'longitude']]))
    df['geo_cluster'] = ['zone_' + str(l) for l in labels]


# --- Step 3: Drop unnecessary columns and keep only new feature and the rest
DROP_COLS = [
    'availability_365',          # same with availability_rate
    'latitude', 'longitude',     # dist_to_cbd + geo_cluster
    'host_since', 'last_review', 'first review'  # feature engine done
]
for df in [train_data, test_data]:
    df.drop(columns=[c for c in DROP_COLS if c in df.columns], inplace=True)

NEW_FEATURES = [
    'amenities_count', 'host_experience_days', 'days_since_last_review', 'is_commercial_host', 'beds_per_person',
    'bath_to_bedroom_ratio', 'accommodates_per_bedroom', 'availability_rate',
    'dist_to_cbd', 'geo_cluster',
]
```

### **Comments:**

Ten new features were constructed from the raw data to capture additional signal
not directly available in the original columns. All transformations were applied
consistently to both the training and test sets. Where a transformation requires
fitting (e.g. K-Means clustering, StandardScaler), parameters were estimated on
the training set only and then applied to the test set to prevent data leakage.


**1. `amenities_count`**
- **Construction:** Count the number of commas in the `amenities` JSON string and
  add 1; assign 0 for empty lists.
- **Justification:** Listings with more amenities offer greater comfort and
  convenience, which typically commands a higher price.


**2. `host_experience_days`**
- **Construction:** Number of days elapsed between `host_since` and the snapshot
  date (2025-12-01), clipped at 0.
- **Justification:** More experienced hosts understand the market better and tend
  to price their listings more competitively and accurately.


**3. `days_since_last_review`**
- **Construction:** Number of days elapsed between `last_review` and the snapshot
  date (2025-12-01).
- **Justification:** Recently reviewed listings signal active demand. Listings
  with no recent reviews may be underpriced or less popular, which affects price.


**4. `is_commercial_host`**
- **Construction:** Binary flag — 1 if `host_listings_count > 1`, else 0.
- **Justification:** Commercial hosts managing multiple listings tend to price
  more strategically and maintain higher quality standards, which may correlate
  with higher prices.


**5. `beds_per_person`**
- **Construction:** `beds / accommodates`. Zero denominator is replaced with NaN
  and imputed in the missing value step.
- **Justification:** A higher bed-to-guest ratio indicates greater comfort.
  Listings with more beds per person can justify higher prices.


**6. `bath_to_bedroom_ratio`**
- **Construction:** `bathrooms / bedrooms`. Zero denominator is replaced with NaN
  and imputed in the missing value step.
- **Justification:** More bathrooms relative to bedrooms is a proxy for luxury.
  Higher ratios are associated with premium listings.


**7. `accommodates_per_bedroom`**
- **Construction:** `accommodates / bedrooms`. Zero denominator is replaced with
  NaN and imputed in the missing value step.
- **Justification:** Measures space density — lower values indicate more spacious
  listings, which typically command higher prices per guest.


**8. `availability_rate`**
- **Construction:** `availability_365 / 365`, giving a proportion in [0, 1].
- **Justification:** Low availability may indicate high demand, which supports
  higher pricing. Highly available listings may be less sought-after and priced
  lower to attract bookings.


**9. `dist_to_cbd`**
- **Construction:** Haversine great-circle distance (km) from the listing's
  coordinates to Brisbane CBD (−27.4698°, 153.0251°).
- **Justification:** Proximity to the CBD is a key price driver in urban markets.
  Listings closer to the city centre typically command a premium due to
  convenience and accessibility.


**10. `geo_cluster`**
- **Construction:** K-Means clustering (k = 5) applied to StandardScaler-
  normalised latitude and longitude. The scaler and cluster centroids were fit on
  the training set only; the test set was transformed using the same parameters.
- **Justification:** Captures spatial price patterns beyond a single distance
  metric. Neighbourhoods within the same zone share similar price levels,
  providing richer location signal than `dist_to_cbd` alone.

listing was last reviewed. This correctly signals that these listings are the
least recently reviewed — more extreme than any observed value would be
inappropriate as it could extrapolate beyond the data.

- `host_is_superhost` → filled with `'f'` (False): a missing superhost status is

- `host_identity_verified` → filled with `'f'` (False): same reasoning — verified

**a. Model Explaination**

We trained and compared three models as our initial set: OLS (standard linear regression), Lasso (regularised linear regression), and Random Forest. The improvement is making a room for more powerful ensemble models such as XGBoost and CatBoost. 
   
- `The OLS` fits a straight line through the data by finding the combination of feature weights that minimises prediction error. It is simple, fast, and easy to interpret. We used it as a baseline to understand how much of the price variation can be explained by a purely linear relationship. Its main weakness here is that price depends on combinations of features (e.g. location × room type) that a straight line cannot capture.. However, the OLS cannot be used to model data with non linear relationship (_Hastie et al., 2017_)

    $$
    \hat{y} = w_1 x_1 + w_2 x_2 + \dots + w_p x_p + b = \mathbf{w}^\top \mathbf{x} + b
    $$

- Lasso works the same way as OLS but adds a penalty for large coefficients. In `Lasso`, we were tunning parameter alpha is first set by CV, and then L1 regularization is performed. The lasso model is a extending method based on Linear regression by adding a penalty part, aiming to prevent model from overfitting that OLS usually did. Their formular of prediction is still the same as above, but following by an optimized equation. The strength of this penalty is controlled by a value called alpha, which we tuned using cross-validation. Lasso improves on OLS when some features are redundant, but still cannot capture non-linear patterns.
    $$
    \min_{\mathbf{w}} \; \sum_{i=1}^{n} \left( y_i - \hat{y}_i \right)^2 + \alpha \sum_{j=1}^{p} |w_j|
    $$

- The tree-based model `RandomForest`takes a completely different approach — it builds many decision trees, each trained on a random subset of the data and features, then averages their predictions. Unlike the two linear models, it can naturally pick up non-linear relationships (e.g. the sharp price jump for waterfront properties) without requiring log-transforms on every feature. The trade-off is that it is harder to interpret and more prone to overfitting on training data (as we will observe below).


**Key difference**
| | OLS | Lasso | Random Forest |
|---|---|---|---|
| **Relationship type** | Linear only | Linear only | Linear & non-linear |
| **Feature selection** | No | Yes (shrinks to zero) | No (but uses feature importance) |
| **Overfitting risk** | Low | Low | Higher on training data |
| **Interpretability** | High | High | Lower |

**b. COMPARE BASE MODEL: _Linear regression with OLS vs Lasso vs Random Forest_**

**Model performance and 5-fold cross validation given default hyperparameter**

```python
from sklearn.linear_model import LinearRegression, Lasso, LassoCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_percentage_error, r2_score
from sklearn.model_selection import cross_validate
import warnings
warnings.filterwarnings('ignore')

# 1. OLS 
lr_ols = LinearRegression()
lr_ols.fit(X_train, y_train)
y_pred_ols_train = lr_ols.predict(X_train)

cv_ols = cross_validate(
    LinearRegression(), X_train, y_train, cv=5,
    scoring=['neg_mean_absolute_percentage_error', 'r2']
)

# 2. Lasso (alpha tuned by LassoCV) - penalize L1 
lasso_cv = LassoCV(cv=5, random_state=42, max_iter=20000)
lasso_cv.fit(X_train, y_train)
best_alpha = lasso_cv.alpha_

lasso = Lasso(alpha=best_alpha, max_iter=20000, random_state=42)
lasso.fit(X_train, y_train)
y_pred_lasso_train = lasso.predict(X_train)

cv_lasso = cross_validate(
    Lasso(alpha=best_alpha, max_iter=20000, random_state=42),
    X_train, y_train, cv=5,
    scoring=['neg_mean_absolute_percentage_error', 'r2']
)

# 3. Random Forest Regression (default baseline)
rf_base = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf_base.fit(X_train, y_train)
y_pred_rf_train = rf_base.predict(X_train)

cv_rf = cross_validate(
    RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1),
    X_train, y_train, cv=5,
    scoring=['neg_mean_absolute_percentage_error', 'r2']
)

# 4. Comparison table
results = {}
for name, y_pred, cv_res in [
    ('OLS',           y_pred_ols_train,   cv_ols),
    ('Lasso',         y_pred_lasso_train, cv_lasso),
    ('Random Forest', y_pred_rf_train,    cv_rf),
]:
    results[name] = {
        'Train MAPE': round(mean_absolute_percentage_error(y_train, y_pred), 4),
        'CV MAPE'   : round(-cv_res['test_neg_mean_absolute_percentage_error'].mean(), 4),
        'Train R²'  : round(r2_score(y_train, y_pred), 4),
        'CV R²'     : round(cv_res['test_r2'].mean(), 4),
    }

print('Base Model Comparison — OLS vs Lasso vs Random Forest')
display(pd.DataFrame(results).T)
print(f'\nLasso best alpha (LassoCV) : {best_alpha:.6f}')

```

#### 2. Model Evaluation Strategy 
<hr style="width:25%;margin-left:0;"> 

`a. Cross-validation:`

To ensure accuracy and prevent overfitting on test set. 5-fold cross-validation will be used:
- The training set (3,735 rows) is split into 5 folds; each fold will utilize 4 parts for training and one part for testing. 
- In each sequequent run, a different part was used for testing, while the remaing 4 part were used for training.
- The final error was calculated as the average of the errors obtained from all 5 runs.

`b. Evaluation metric:`

- We use two metrics **_Mean Absolute Percentage Error (MAPE)_**: _The average size of the prediction error expressed as a percentage of the true value_ and **_R_squared score_**: _The proportion of variance in the target that the model explains_ on log-transformed price, consistent with the Kaggle competition metric. Futhermore, we can also use a diverse range of evaluation metrics to allow for a comprehensive evalution. 

The three models also show distinct bias–variance behaviour Base on evaluation metric:

- Lasso underfits: its training and cross-validated R² are both low (≈0.47–0.48), indicating that the default regularisation is too strong and suppresses useful signal — it is the weakest model. 
- OLS is a stable, well-generalised baseline, with a very small train–CV gap (R² 0.7161 → 0.6956), but it is limited to linear relationships and therefore cannot capture the non-linear.
- Random Forest shows the opposite profile: a large train–CV gap (R² 0.9643 → 0.7405) that signals overfitting, yet it still delivers the best generalisation of the three because it captures the non-linear interactions the linear models miss.

- Using both together gives a clearer picture where MAPE shows raw prediction accuracy, while R² shows how well the model captures the overall pattern in the data.

`c. Comparison:`

In overall, models are ranked by mean CV MAPE (lower = better). Standard deviation across folds will also be reported to assess stability. The best-performing model will be retrained on the full training AND test set for Kaggle submission. #### 3. Model Training and Hyperparameter Tuning 

<hr style="width:25%;margin-left:0;"> 

**a. Pipeline Hyperparameter tunning for 3 models**

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, cross_validate
from sklearn.metrics import mean_absolute_percentage_error, r2_score
import warnings
warnings.filterwarnings('ignore')

# Pipeline definition creation
'''
Linear models: RobustScaler (resistant to outliers) + model
Random Forest: no scaling needed (tree splits are scale-invariant)`
'''
ols_pipe = Pipeline([('scaler', RobustScaler()), ('model', LinearRegression())]) # OLS pipline
lasso_pipe = Pipeline([('scaler', RobustScaler()), ('model', Lasso(max_iter=20000))]) # Lasso pipiline
rf_pipe = Pipeline([('model', RandomForestRegressor(random_state=42, n_jobs=-1))]) # Random Forest pipiline

# Create hyperparameter grids
ols_grid = {'model__fit_intercept': [True, False]}

lasso_grid = {'model__alpha': np.logspace(-4, 1, 40).tolist()}

rf_grid = {
    'model__n_estimators'  : [100, 300, 500],
    'model__max_depth'     : [10, 20, None],
    'model__min_samples_leaf': [1, 4],
    'model__max_features'  : ['sqrt', 0.5],
}

# GridSearchCV (5-fold, scored by MAPE)
searches = {
    'OLS'          : GridSearchCV(ols_pipe,   ols_grid,   cv=5, scoring='neg_mean_absolute_percentage_error', n_jobs=-1),
    'Lasso'        : GridSearchCV(lasso_pipe, lasso_grid, cv=5, scoring='neg_mean_absolute_percentage_error', n_jobs=-1),
    'Random Forest': GridSearchCV(rf_pipe,    rf_grid,    cv=5, scoring='neg_mean_absolute_percentage_error', n_jobs=-1),
}

for name, gs in searches.items():
    gs.fit(X_train, y_train)
    print(f'{name}')
    print(f'  Best params : {gs.best_params_}')
    print(f'  Best CV MAPE: {-gs.best_score_:.4f}\n')

# Performance table: train vs CV (MAPE & R²)
tuned_results = {}
best_estimators = {}
for name, gs in searches.items():
    best = gs.best_estimator_
    best_estimators[name] = best
    y_pred_train = best.predict(X_train)
    cv_res = cross_validate(
        best, X_train, y_train, cv=5,
        scoring=['neg_mean_absolute_percentage_error', 'r2']
    )
    tuned_results[name] = {
        'Train MAPE': round(mean_absolute_percentage_error(y_train, y_pred_train), 4),
        'CV MAPE'   : round(-cv_res['test_neg_mean_absolute_percentage_error'].mean(), 4),
        'Gap (overfit?)': round(
            mean_absolute_percentage_error(y_train, y_pred_train)
            - (-cv_res['test_neg_mean_absolute_percentage_error'].mean()), 4),
        'Train R²'  : round(r2_score(y_train, y_pred_train), 4),
        'CV R²'     : round(cv_res['test_r2'].mean(), 4),
    }

results_df = pd.DataFrame(tuned_results).T
print('Tuned Model Comparison — MAPE & R² (5-fold CV)')
display(results_df)
print('Note: Gap = Train MAPE − CV MAPE. Negative gap → overfitting (train much better than CV).')
print('      Gap near 0 → good generalisation. Positive gap → underfitting.')
```

**Comment and interpretation after tunning hyperparameter**

**_OLS (Linear Regression)_**

OLS has essentially no meaningful hyperparameters to tune. The training and CV MAPE are close to each other (gap ≈ 0), which confirms that OLS is not overfitting but show the sign of underfitting because the model is too simple to capture all the complexity of target variable price relationship. 

However, the CV R² of 0.69 tells us that OLS can only explain about 70% of the variation in listing price using a linear combination of features. Given that price depends on complex interactions. A private room in a beachfront suburb is not simply "location effect + room type effect" added together — this ceiling is expected. OLS is doing the best it can within its structural limits.

**_Lasso_**

The best alpha found by cross-validation was 0.000, a very small value close to zero. This is a key finding: when alpha is near zero, the L1 penalty is barely active, and Lasso reduces almost entirely to OLS. This explains why both models' CV MAPE and R² are nearly identical.

What this tells us is that in this dataset, the problem is not that we have too many noisy features pulling the model in wrong directions. It is that the relationship between features and price is genuinely non-linear, which no amount of coefficient shrinkage can fix. Lasso's intended benefit (removing weak features) had almost no effect here because most features are contributing some signal, just not in a linear way.

The gap for Lasso (Train MAPE − CV MAPE ≈ 0.00) is similarly small, confirming there is no meaningful overfitting, but both models generalise predictably, just not very accurately.

Random Forest improve but not too significantly

#### 5. Model Improvement and Reflection 

<hr style="width:25%;margin-left:0;"> 

**a. Feature re-selection:**

Our first model used only the original 26 variables. After reviewing which features were driving predictions, we rebuilt the feature set with several targeted improvements:

1. `Amenities`: 

- The raw amenities column contains a free-text list with over 2,000 unique items. Rather than just counting the total number of amenities, we extracted the top 25 most frequently listed and price-relevant amenities as individual binary columns (1 = present, 0 = absent). This allows the model to distinguish, for example, that a listing with a pool commands a higher price — a signal that a plain count misses entirely.

2. `Listing name keywords`

- We extracted keyword-based binary flags from the listing name (e.g. "pool", "ocean view", "studio") to capture host-marketed features not found elsewhere in the structured data.

3. `Location`:

- Raw latitude and longitude capture location basically but no reference, but not meaningfully. We replaced them with two purpose-built variables: 
    - `dist_to_cbd` (straight-line distance to Brisbane CBD)
    - `neighbourhood_te` (target-encoded suburb price signal). KMeans clustering was also applied to group listings into geographic zones, fitted on training data only to prevent leakage.

5. `Neighbourhood`: 

- Instead of collapsing 95+ suburbs into just "top 5 + other" (which throws away most geographic information), we switched to target encoding using KFold — a leakage-free method that maps each suburb to a price signal based on the training data.

<hr style="width:25%;margin-left:0;">  
<hr style="width:25%;margin-left:0;"> 

**b. Tunning hyperparameter**

- Given that price data contains outliers and non-linear patterns, the priority was not maximising fit on the training set but closing the gap between training and cross-validation error. Tuning regularisation-related parameters. We mannually set low range of depth limits and leaf size to directly addresses the overfitting observed in the initial model.

**Random Forest Grid — Justification (overfitting-focused)**

| Hyperparameter | Values | Role | Why chosen (overfitting context) |
|---|---|---|---|
| `n_estimators` | 100, 300, 500 | Number of trees | More trees reduce variance and stabilise predictions; does not itself cause overfitting (bagging averages over trees). |
| `max_depth` | 10, 20, None | Tree depth | Main lever against overfitting, capping depth stops fully-grown trees from memorising the training data, aiming to minimize the max depth |
| `min_samples_leaf` | 1, 4 | Min samples per leaf | Larger leaves smooth predictions and resist price outliers, narrowing the train–CV gap. |
| `max_features` | 'sqrt', 0.5 | Features per split | Fewer features decorrelate the trees, increasing diversity and lowering variance. |

#### **Justify why these varibles is effective**

Comparing the top-20 feature importances before and after re-engineering (see charts below) shows a clear shift:

- neighbourhood_te (rank 6): the target-encoded location carries genuine signal.
- has_pool (rank 16): an amenity binary makes the list, showing that extracting individual binaries captures value that amenities_count alone cannot.
- bath_to_bedroom_ratio (rank 18): the engineered ratio is being used by the model.
- host_acceptance_rate (rank 19): usable once converted from a percentage string to a numeric value.

→ This is direct evidence that the feature re-selection produced usable signal. After re-selection, the engineered features all appear in the top-20 MDI, indicating that they contribute meaningfully to the model's decisions. Importance is also more evenly distributed (the leading feature falls from 0.30 to 0.175), reducing the model's reliance on any single variable. That said, MDI only reflects how much a feature is used, not how well the model predicts; the effectiveness of the improvement is confirmed by the average score achieved on the Kaggle test-set submission.

- Features that possibly drive price (feature importance): size and location remain at the top in both versions (room_type, accommodates, bedrooms, bathrooms, dist_to_cbd) — a consistent and credible result. These variables can serve as a basis for decision-making, and warrant deeper exploration to extract additional predictive gains.


**c. Main limitation preventing a higher ranking**

- **_Persistent overfitting:_** Despite tuning, our training MAPE remains noticeably lower than the cross-validated error. Tree-based models learn the training listings in fine detail, and with only ~3,700 training rows, some of that detail is noise rather than real pattern. Stronger regularisation (shallower trees, larger minimum leaf size) partially addresses this but cannot eliminate it entirely.

- **_Review text was not used:_** A top ranking might require richer inputs in reality, for example, we could have using sentimental analysis to analyse text and description of listings. This was a conscious trade-off given lack of in-depth understanding but represents a meaningful missed signal.
 And some significant correlation between features and price we're missing during the feature prrocess.

- **_Neighbourhood encoding ceiling_:** Even with KFold target encoding, suburbs with very few listings produce unreliable encoded values. Sparse suburbs effectively fall back to the global mean, losing local price information. So that we are unable to capture the full feature relationship in-depth.

**c. Extension model set: XGBoost and CatBoost**

```python
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
# import catboost
from catboost import CatBoostRegressor
from sklearn.metrics import mean_squared_error
```

```python
# 1. BASELINE BENCHMARK
from sklearn.model_selection import RandomizedSearchCV, cross_val_score

print("Baseline CV MAPE (5-fold):")
baseline = {
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42),
    'XGBoost'      : XGBRegressor(n_estimators=200, random_state=42, verbosity=0),
    'CatBoost'   : CatBoostRegressor(n_estimators=200, random_state=42, verbose=0)
}
for name, model in baseline.items():
    scores = cross_val_score(model, X_train, y_train, cv=5,
                             scoring='neg_mean_absolute_percentage_error', n_jobs=-1)
    print(f"  {name}: {-scores.mean():.4f} ± {scores.std():.4f}")

# 2. HYPERPARAMETER TUNING
tree_param_distributions = {
    'n_estimators'    : [100, 200, 300, 500],
    'max_depth'       : [10, 20, 30, None],
    'max_features'    : ['sqrt', 'log2', 0.5, 0.8],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4, 8]
}

# Expanded XGBoost grid — more trees + regularisation params
xgb_gs = RandomizedSearchCV(
    XGBRegressor(objective='reg:absoluteerror', eval_metric='mape', random_state=42, verbosity=0, tree_method='hist'),
    param_distributions={
        'n_estimators'    : [300, 500, 1000, 2000],
        'max_depth'       : [3, 4, 6, 8],
        'eta'             : [0.005, 0.01, 0.05, 0.1],
        'subsample'       : [0.6, 0.7, 0.8, 0.9, 1.0],
        'colsample_bytree': [0.5, 0.7, 0.8, 1.0],
        'min_child_weight': [1, 3, 5, 10],
        'reg_alpha'       : [0, 0.01, 0.1, 1.0],
        'reg_lambda'      : [0.5, 1.0, 2.0, 5.0],
    },
    n_iter=30, cv=5, scoring='neg_mean_absolute_percentage_error',
    n_jobs=-1, random_state=42
)

rf_rs = RandomizedSearchCV(
    RandomForestRegressor(random_state=42, criterion='absolute_error'),
    param_distributions=tree_param_distributions,
    n_iter=20, cv=5, scoring='neg_mean_absolute_percentage_error', n_jobs=-1, random_state=42
)

cat_rs = RandomizedSearchCV(
    CatBoostRegressor(loss_function='MAPE', random_state=42, verbose=0),
    param_distributions={
        'depth'          : [4, 6, 8, 10],
        'l2_leaf_reg'    : [1, 3, 5, 7, 9],
        'learning_rate'  : [0.01, 0.03, 0.05, 0.1, 0.2],
        'iterations'     : [200, 300, 500, 1000]
    },
    n_iter=20, cv=5, scoring='neg_mean_absolute_percentage_error', n_jobs=-1, random_state=42
)

for search in [xgb_gs, rf_rs, cat_rs]:
    search.fit(X_train, y_train)

print("\nTuning Results:")

print("Model: XGBooost\n")
print(f"Best Params: {xgb_gs.best_params_}\n")
print(f" CV MAPE: {-xgb_gs.best_score_:.4f}\n")

print("Model: Random Forest\n")
print(f"Best Params: {rf_rs.best_params_}\n")
print(f" CV MAPE: {-rf_rs.best_score_:.4f}\n")

print("Model: CatBoost\n")
print(f"Best Params: {cat_rs.best_params_}\n")
print(f" CV MAPE: {-cat_rs.best_score_:.4f}\n")
```
